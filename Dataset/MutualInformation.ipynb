```python
# We will calculate the mutual information between each feature column and the target column
# Because the variables are continuo, we will split them into intervals to make them discrete
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('cleaned_viral_songs.csv')
```

```python
df.head()
```

```python
features = df.drop(columns=['viral'])
target = df['viral']
mutualInformation = {}
numBins = 100
```

```python
for column in features.columns:
    binnedFeature = pd.qcut(df[column], q=numBins, duplicates='drop')

    nTotal = len(binnedFeature)

    pX = binnedFeature.value_counts() / nTotal
    pY = target.value_counts() / nTotal

    jointCounts = pd.crosstab(binnedFeature, target)
    pXY = jointCounts / nTotal

    mi = 0.0

    for x in pXY.index:
        for y in pXY.columns:
            jointProb = pXY.loc[x,y]
            if jointProb > 0:
                marginalProduct = pX[x] * pY[y]
                mi += jointProb * np.log2(jointProb / marginalProduct)

    mutualInformation[column] = mi
```

```python
sortedMI = sorted(mutualInformation.items(), key=lambda item: item[1], reverse=True)
```

```python
print("Mutual Information Scores (in bits):")
print("-" * 40)
for featureName, score in sortedMI:
    print(f"{featureName:<20}: {score:.4f} bits")
```
